
# DATA QUALITY ASSESSMENT


```python
import pandas as pd
import numpy as np
```

```python
# ============================================================
# 1. LOAD DATASET
# ============================================================

df=pd.read_excel(r"C:\Users\Dell\Desktop\Pricing Strategy & Bundling Analytics\Dataset\online_retail_II.xlsx")
```

```python
df.head()
```

```python
# ============================================================
# 2. INITIAL DATA INSPECTION
# ============================================================

print("\n--- FIRST 5 ROWS ---")
display(df.head())

print("\n--- RANDOM SAMPLE ---")
display(df.sample(5, random_state=42))

print("\n--- DATASET SHAPE ---")
print(df.shape)

print("\n--- COLUMN NAMES ---")
print(df.columns.tolist())

print("\n--- DATA TYPES ---")
df.info()
```

```python
# ============================================================
# 3. STATISTICAL SUMMARY
# ============================================================

print("\n--- NUMERICAL SUMMARY ---")
display(df.describe().T)

print("\n--- CATEGORICAL SUMMARY ---")
display(df.describe(include="object").T)
```

```python
# ============================================================
# 4. MISSING VALUE ANALYSIS
# ============================================================

missing_report = pd.DataFrame({
    "Column": df.columns,
    "Missing_Count": df.isna().sum().values,
    "Missing_Percentage":
        ((df.isna().sum().values*100)/len(df)).round(2)
})

missing_report = missing_report.sort_values(
    by="Missing_Percentage",
    ascending=False
)

print("--- MISSING VALUE REPORT ---")
display(missing_report)
```

```python
# ============================================================
# 5. DUPLICATE ANALYSIS
# ============================================================

duplicate_count = df.duplicated().sum()
duplicate_percentage = duplicate_count / len(df) * 100

print(f"Duplicate Rows       : {duplicate_count:,}")
print(f"Duplicate Percentage : {duplicate_percentage:.2f}%")

if duplicate_count > 0:
    print("\n--- SAMPLE DUPLICATE ROWS ---")
    display(df[df.duplicated(keep=False)].head(10))
```

```python
# ============================================================
# 6. CANCELLATION / RETURN ANALYSIS
# ============================================================

df["Invoice"] = df["Invoice"].astype(str)

cancellation_mask = df["Invoice"].str.startswith("C")

cancellation_count = cancellation_mask.sum()
cancellation_percentage = cancellation_count / len(df) * 100

print(f"\nCancellation Rows       : {cancellation_count:,}")
print(f"Cancellation Percentage : {cancellation_percentage:.2f}%")

print("\n--- SAMPLE CANCELLATIONS ---")
display(df[cancellation_mask].head(10))
```

```python
# ============================================================
# 7. QUANTITY VALIDATION
# ============================================================

negative_quantity_count = (df["Quantity"] < 0).sum()
zero_quantity_count = (df["Quantity"] == 0).sum()

print("\n--- QUANTITY VALIDATION ---")
print(f"Negative Quantity Rows : {negative_quantity_count:,}")
print(f"Zero Quantity Rows     : {zero_quantity_count:,}")

display(df["Quantity"].describe())
```

```python
# ============================================================
# 8. PRICE VALIDATION
# ============================================================

negative_price_count = (df["Price"] < 0).sum()
zero_price_count = (df["Price"] == 0).sum()

print("\n--- PRICE VALIDATION ---")
print(f"Negative Price Rows : {negative_price_count:,}")
print(f"Zero Price Rows     : {zero_price_count:,}")

display(df["Price"].describe())

print("\n--- INVALID PRICE EXAMPLES ---")

invalid_price_mask = df["Price"] <= 0

display(df[invalid_price_mask].head(10))
```

```python
# ============================================================
# 9. CUSTOMER ID ANALYSIS
# ============================================================

missing_customer_count = df["Customer ID"].isna().sum()
missing_customer_percentage = (
    (df["Customer ID"].isna().sum() * 100)/len(df)
)

print("\n--- CUSTOMER ANALYSIS ---")
print(f"Missing Customer IDs  : {missing_customer_count:,}")
print(f"Missing Customer ID % : {missing_customer_percentage:.2f}%")
print(f"Unique Customers      : {df['Customer ID'].nunique():,}")
```

```python
# ============================================================
# 10. DATE VALIDATION
# ============================================================

df["InvoiceDate"] = pd.to_datetime(
    df["InvoiceDate"],
    errors="coerce"
)

missing_dates = df["InvoiceDate"].isna().sum()

print("\n--- DATE VALIDATION ---")
print(f"Missing/Invalid Dates : {missing_dates:,}")
print(f"Minimum Date          : {df['InvoiceDate'].min()}")
print(f"Maximum Date          : {df['InvoiceDate'].max()}")
```

```python
# ============================================================
# 11. PRODUCT ANALYSIS
# ============================================================

print("\n--- PRODUCT ANALYSIS ---")
print(f"Unique Products      : {df['StockCode'].nunique():,}")
print(f"Unique Descriptions  : {df['Description'].nunique():,}")
print(f"Missing Descriptions : {df['Description'].isna().sum():,}")
```

```python
# ============================================================
# 12. COUNTRY ANALYSIS
# ============================================================

print("\n--- COUNTRY ANALYSIS ---")
print(f"Unique Countries : {df['Country'].nunique():,}")

country_summary = (
    df["Country"]
    .value_counts()
    .head(15)
    .reset_index()
)
  
country_summary.columns =[
    "Country",
    "Transaction_Rows"
]

display(country_summary)
```

```python
# ============================================================
# 13. OVERALL DATA QUALITY REPORT
# ============================================================

data_quality_report = pd.DataFrame({
    "Column": df.columns,
    "Data_Type": df.dtypes.astype(str   ).values,
    "Missing_Count": df.isna().sum().values,
    "Missing_%": (df.isna().mean() * 100).round(2).values,
    "Unique_Values": df.nunique().values
})

data_quality_report["Quality_Status"] = np.where(
    data_quality_report["Missing_Count"] == 0,
    "PASS",
    "REVIEW"
)

data_quality_report = data_quality_report.sort_values(
    "Missing_%",
    ascending=False
)

print("\n--- OVERALL DATA QUALITY REPORT ---")
display(data_quality_report)
```

```python

# ============================================================
# 14. PROJECT-LEVEL SUMMARY
# ============================================================

summary = {
    "Total Rows": len(df),
    "Total Columns": len(df.columns),
    "Duplicate Rows": duplicate_count,
    "Cancellation Rows": cancellation_count,
    "Negative Quantity Rows": negative_quantity_count,
    "Zero Quantity Rows": zero_quantity_count,
    "Negative Price Rows": negative_price_count,
    "Zero Price Rows": zero_price_count,
    "Missing Customer IDs": missing_customer_count,
    "Missing Dates": missing_dates,
    "Unique Products": df["StockCode"].nunique(),
    "Unique Customers": df["Customer ID"].nunique(),
    "Unique Countries": df["Country"].nunique()
}

summary_report = pd.DataFrame(
    summary.items(),
    columns=["Metric", "Value"]
)

print("\n--- PROJECT DATA QUALITY SUMMARY ---")
display(summary_report)
```

```python
# ============================================================
# 15. SAVE REPORTS
# ============================================================

data_quality_report.to_csv(
    "data_quality_report.csv",
    index=False
)

summary_report.to_csv(
    "data_quality_summary.csv",
    index=False
)

print("\nData quality reports saved successfully.")
```
