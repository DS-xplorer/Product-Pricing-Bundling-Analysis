
# STRATEGIC PRICING ANALYSIS


```python
import pandas as pd
import numpy as np
```

```python
product_elasticity=pd.read_csv("product_elasticity.csv")
```

```python
# ============================================================
# STEP 1 — ELASTICITY SUMMARY
# ============================================================

elasticity_summary = (
    product_elasticity
    .groupby("Elasticity_Type")
    .agg(
        Products=("Stock_Code", "nunique"),
        Total_Revenue=("Total_Revenue", "sum"),
        Total_Quantity=("Total_Quantity", "sum")
    )
    .reset_index()
)

display(elasticity_summary)
```

```python
# ============================================================
# HIGH-REVENUE ELASTIC PRODUCTS
# ============================================================

high_revenue_elastic = product_elasticity[
    product_elasticity["Elasticity_Type"] == "Elastic"
].copy()

high_revenue_elastic = high_revenue_elastic.sort_values(
    "Total_Revenue",
    ascending=False
)

display(
    high_revenue_elastic[
        [
            "Stock_Code",
            "Average_Elasticity",
            "Average_Price",
            "Total_Quantity",
            "Total_Revenue",
            "Number_of_Months",
            "Elasticity_Type"
        ]
    ].head(20)
)
```

```python
# ============================================================
# HIGH-REVENUE INELASTIC PRODUCTS
# ============================================================

high_revenue_inelastic = product_elasticity[
    product_elasticity["Elasticity_Type"] == "Inelastic"
].copy()

high_revenue_inelastic = high_revenue_inelastic.sort_values(
    "Total_Revenue",
    ascending=False
)

display(
    high_revenue_inelastic[
        [
            "Stock_Code",
            "Average_Elasticity",
            "Average_Price",
            "Total_Quantity",
            "Total_Revenue",
            "Number_of_Months",
            "Elasticity_Type"
        ]
    ].head(20)
)
```

```python
# ============================================================
# PRICING OPPORTUNITY MATRIX
# ============================================================

product_elasticity["Pricing_Strategy"] = np.where(
    product_elasticity["Elasticity_Type"] == "Elastic",
    "Protect Price / Targeted Discount",
    np.where(
        product_elasticity["Elasticity_Type"] == "Inelastic",
        "Test Moderate Price Increase",
        "Review Further"
    )
)

display(
    product_elasticity[
        [
            "Stock_Code",
            "Average_Elasticity",
            "Average_Price",
            "Total_Quantity",
            "Total_Revenue",
            "Number_of_Months",
            "Elasticity_Type",
            "Pricing_Strategy"
        ]
    ].head(20)
)
```

```python
# ============================================================
# DAY 6 — STEP 5
# TOP PRICING OPPORTUNITIES
# ============================================================

reliable_products = product_elasticity[
    product_elasticity["Number_of_Months"] >= 4
].copy()

print(f"Reliable products: {len(reliable_products):,}")

display(
    reliable_products[
        [
            "Stock_Code",
            "Average_Elasticity",
            "Average_Price",
            "Total_Quantity",
            "Total_Revenue",
            "Number_of_Months",
            "Elasticity_Type",
            "Pricing_Strategy"
        ]
    ]
    .sort_values("Total_Revenue", ascending=False)
    .head(20)
)
```

```python
# ============================================================
# DAY 6 — STEP 6
# HIGH-VALUE PRICING SEGMENTS
# ============================================================

# Find the median revenue among reliable products
revenue_threshold = reliable_products["Total_Revenue"].median()

print(f"Revenue threshold: £{revenue_threshold:,.2f}")

# Create revenue segment
reliable_products["Revenue_Segment"] = np.where(
    reliable_products["Total_Revenue"] >= revenue_threshold,
    "High Revenue",
    "Lower Revenue"
)

# Display the important combinations
pricing_opportunities = (
    reliable_products[
        reliable_products["Revenue_Segment"] == "High Revenue"
    ]
    .sort_values("Total_Revenue", ascending=False)
)

display(
    pricing_opportunities[
        [
            "Stock_Code",
            "Average_Elasticity",
            "Average_Price",
            "Total_Quantity",
            "Total_Revenue",
            "Number_of_Months",
            "Elasticity_Type",
            "Revenue_Segment",
            "Pricing_Strategy"
        ]
    ].head(20)
)
```

```python
# ============================================================
# DAY 6 — STEP 7
# FINAL PRICING OPPORTUNITY TABLE
# ============================================================

final_pricing_opportunities = pricing_opportunities[
    pricing_opportunities["Elasticity_Type"].isin(
        ["Elastic", "Inelastic"]
    )
].copy()

# Create business priority
final_pricing_opportunities["Priority"] = np.where(
    final_pricing_opportunities["Elasticity_Type"] == "Inelastic",
    "High",
    "Medium"
)

# Sort by priority and revenue
final_pricing_opportunities = (
    final_pricing_opportunities
    .sort_values(
        ["Priority", "Total_Revenue"],
        ascending=[True, False]
    )
)

display(
    final_pricing_opportunities[
        [
            "Stock_Code",
            "Average_Elasticity",
            "Average_Price",
            "Total_Quantity",
            "Total_Revenue",
            "Number_of_Months",
            "Elasticity_Type",
            "Pricing_Strategy",
            "Priority"
        ]
    ].head(20)
)
```

```python
# Check high-revenue elastic products
high_revenue_elastic = pricing_opportunities[
    pricing_opportunities["Elasticity_Type"] == "Elastic"
].copy()

display(
    high_revenue_elastic[
        [
            "Stock_Code",
            "Average_Elasticity",
            "Average_Price",
            "Total_Quantity",
            "Total_Revenue",
            "Number_of_Months",
            "Elasticity_Type",
            "Pricing_Strategy"
        ]
    ]
    .sort_values("Total_Revenue", ascending=False)
    .head(20)
)
```

```python
# Add Pricing Strategy

product_elasticity["Pricing_Strategy"] = np.where(
    product_elasticity["Elasticity_Type"] == "Elastic",
    "Protect Price / Targeted Discount",
    np.where(
        product_elasticity["Elasticity_Type"] == "Inelastic",
        "Test Moderate Price Increase",
        "Review Further"
    )
)

# Update the existing CSV file
product_elasticity.to_csv(
    "product_elasticity.csv",
    index=False
)

print("product_elasticity.csv updated successfully! ✅")
```

```python
# ============================================================
# FINAL PRICING OPPORTUNITIES
# ============================================================

reliable_products = product_elasticity[
    product_elasticity["Number_of_Months"] >= 4
].copy()

reliable_products["Priority"] = np.where(
    (
        (reliable_products["Elasticity_Type"] == "Elastic") &
        (reliable_products["Total_Revenue"] >= reliable_products["Total_Revenue"].median())
    ),
    "High",
    np.where(
        reliable_products["Elasticity_Type"] == "Inelastic",
        "Medium",
        "Review"
    )
)

pricing_opportunities = reliable_products[
    [
        "Stock_Code",
        "Average_Elasticity",
        "Average_Price",
        "Total_Quantity",
        "Total_Revenue",
        "Number_of_Months",
        "Elasticity_Type",
        "Pricing_Strategy",
        "Priority"
    ]
].sort_values(
    "Total_Revenue",
    ascending=False
)

display(pricing_opportunities.head(20))

pricing_opportunities.to_csv(
    "final_pricing_opportunities.csv",
    index=False
)

print("final_pricing_opportunities.csv saved successfully!")
```
